# 2. test dataset

```python
test=pd.read_csv(r"C:\Users\KIIT\Desktop\eLabs\Titanic\titanic\test.csv")
test
```

```python
test.head()
```

```python
test.info()
```

```python
test.describe()
```

```python
#checking null values 
test.isnull().sum()
```

```python
#drop cabin as too much null values present in it
test.drop("Cabin", axis=1, inplace=True)
```

```python
test["Age"].fillna(test["Age"].median(), inplace=True)
test["Fare"].fillna(test["Fare"].median(), inplace=True)
```

```python
test.isnull().sum()
```

```python
test.dtypes
```

```python
test.head()
```

# EDA Visualization

```python

sns.countplot(x='Pclass', data=test)
plt.title("Passenger Class Distribution")
plt.show()
```

```python

sns.countplot(x='Sex', data=test)
plt.title("Gender Distribution")
plt.show()
```

```python
#age distribution
sns.histplot(test['Age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
#age vs fare
sns.scatterplot(x='Age', y='Fare', data=test)
plt.title("Age vs Fare")
plt.show()
```

```python
#pclass vs fare
sns.boxplot(x='Pclass', y='Fare', data=test)
plt.title("Pclass vs Fare")
plt.show()
```

```python
#heatmap
sns.heatmap(test.corr(numeric_only=True),annot=True,cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```

# 3. Train dataset

```python
train=pd.read_csv(r"C:\Users\KIIT\Desktop\eLabs\Titanic\titanic\train.csv")
train.head()
```

```python
train.describe()
```

```python
train.isnull().sum()
```

```python
train.shape
```

```python
train.drop("Cabin", axis=1, inplace=True)
```

```python
train["Age"].fillna(train["Age"].median(), inplace=True)
train["Embarked"].fillna(train["Embarked"].mode()[0], inplace=True)
```

```python
train.isnull().sum()
```

```python
#eda
sns.countplot(x='Survived', data=train)
plt.title("Survival Distribution")
plt.show()
```

```python
#age distribution
sns.histplot(train['Age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
sns.pairplot(train[['Survived','Age','Fare','Pclass']])

plt.show()
```

```python
sns.boxplot(train['Age'])
plt.title("Age Outliers")
plt.show()
```

```python
sns.heatmap(train.corr(numeric_only=True),annot=True,cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
